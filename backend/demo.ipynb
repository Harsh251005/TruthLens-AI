```python
import tensorflow as tf
import numpy as np

model    = tf.keras.models.load_model('deepfake_detector.keras')
backbone = model.get_layer('efficientnetv2-b2')
last_conv = backbone.get_layer('top_activation')

backbone_grad_model = tf.keras.Model(
    inputs  = backbone.input,
    outputs = [last_conv.output, backbone.output]
)

dummy = np.random.rand(1, 260, 260, 3).astype(np.float32)
conv_outputs, backbone_out = backbone_grad_model(dummy, training=False)
print('conv_outputs shape:', conv_outputs.shape)
print('backbone_out shape:', backbone_out.shape)
```
